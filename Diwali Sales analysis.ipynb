```python
gender_count = sns.countplot( x = 'Gender' , data = df)
```

```python
gender_count = sns.countplot( x = 'Gender' , data = df)
for bar in gender_count.containers:
    gender_count.bar_label(bar)
```

```python
sales_gender = df.groupby( ['Gender'], as_index = False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.barplot(x = 'Gender', y = 'Amount', data = sales_gender)
```

Orders and total amount spend in higher of females then male

```python
sns.countplot(data = df, x = 'Age Group')
```

```python
sns.countplot(data = df, x = 'Age Group', hue = 'Gender')
```

We can see that the females of age group 26-35 is buying the most products

```python
ax = sns.countplot(data = df, x = 'Age Group', hue = 'Gender')
for bar in ax.containers:
    ax.bar_label(bar)
```

```python
sales_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.barplot(x = 'Age Group', y = 'Amount', data = sales_age)
```

```python
df.columns
```

```python
sales_state = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='State', ascending=False)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x = 'State', y = 'Orders', data = sales_state)
```

By the above graph we can see that the most orders are received from the state of Uttar Prades and least orders from Telangana

```python
sales_state = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='State', ascending = False)

sns.set(rc={'figure.figsize':(22,5)})
sns.barplot(x = 'State', y = 'Amount', data = sales_state)
```

```python
ax = sns.countplot( x = 'Marital_Status', data = df)
sns.set({'figure.figsize':(10,5)})
for bar in ax.containers:
    ax.bar_label(bar)
```

```python
sales_status = df.groupby(['Marital_Status', 'Gender'], as_index=False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)
sns.set(rc={'figure.figsize' : (6,5)})
sns.barplot(data = sales_status, x = 'Marital_Status', y = 'Amount', hue = 'Gender')
```

```python
ax = sns.countplot(data = df, x = 'Occupation')
sns.set(rc={'figure.figsize':(23,5)})

for bar in ax.containers:
    ax.bar_label(bar)
```

```python
sales_occ = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by = 'Amount', ascending=False)
sns.barplot(x = 'Occupation', y = 'Amount', data = sales_occ)
```

```python
ax = sns.countplot(data = df, x = 'Product_Category')
sns.set(rc={'figure.figsize':(40,10)})
for bar in ax.containers:
    ax.bar_label(bar)
```

```python
sales_product = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.set(rc={'figure.figsize':(40,10)})
sns.barplot(x = 'Product_Category', y = 'Amount', data = sales_product)
```

Here we can see that the orders are more in clothing & apparel but the amount spend by the buyers is more in food category
